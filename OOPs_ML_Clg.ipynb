OOP (Object-Oriented Programming) is a way of organizing code around real-world objects instead of just a list of instructions.

A Class = the blueprint/design of a car (not an actual car yet, just the plan — engine type, color options, etc.)

An Object = an actual car built FROM that blueprint (you can build many cars from ONE blueprint!)

Each car (object) shares the SAME basic design (class) but can have its own specific details (color, brand, speed).

```python
class Student:
  clg='BIET Lucknow'

  def __init__(self,name,age,course,marks):
    self.name=name
    self.age=age
    self.course=course
    self.marks=marks

  def intro(self):
    print(f"My name is {self.name} .I am {self.year} years old")

  def check_result(self):
    if self.marks > 33:
      print(f"BRO is passed with {self.marks} marks")
    else:
      print(f"BRO is failed by {33-self.marks} marks")

a = 'Ankit',21,'Btech',23
student1=Student(*a)
print(student1.name,student1.age,student1.course,student1.clg)
Student.clg="IIT Ropar"
print(student1.name,student1.age,student1.course,student1.clg)

student1.check_result()
```

1️⃣ Inheritance — A Class "Inheriting" from Another

Inheritance lets one class (child) automatically get all the features of another class (parent), without rewriting that code.

```python
class Car:
  wheel = "four wheeler"
  def __init__(self):
    pass

  def get_wheels():
    return Car.wheel()
  def category(self):
    self.category = 'car'
    return self.category

  def wheel(self):
    self.wheel = 4
    return self.wheel


class Company(Car):
  def __init__(self):
    self.country='India'

maruti=Company()
print(maruti.country,maruti.category(),maruti.wheel(),maruti.get_wheels())
```
