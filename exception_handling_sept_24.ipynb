```python
try:
  a=[1,2,3]
  print(a[10])
except Exception as e:
  print(f"An error occurred:{type(e).__name__}")
  print(f"Error msg:{e}")
```

```python
import csv
students_data = [
    ["name", "age", "course"],
    ["Prakhar", 21, "Machine Learning"],
    ["Anjali", 22, "Data Science"],
    ["Rohit", 20, "Web Development"]
]

with open("Student.csv","w") as file:
  a=csv.writer(file)
  a.writerows(students_data)
```

```python
with open("Student.csv","r") as file:
  b=csv.reader(file)
  next(b)
  for i in b:
    print(f"My name is {i[0]},age is {i[1]} and enroll in this {i[2]} course")
```
